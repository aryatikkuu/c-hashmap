```python
%%writefile main.cpp
#include <iostream>
#include <vector>
#include <string>
using namespace std;
#include <utility>


//jersey_number -> name
//insert/update players
//delete players
//search players

vector<vector <pair <int,string> > > players(10, std ::vector <pair <int,string> > (4));

int hashjersey(int jersey_number){
  return jersey_number % 10;
}

void insert_seperate_chaining(int jersey_number, string name){

  int pos = hashjersey(jersey_number);
  int current_index = 0;

  while (current_index < players[pos].size() && players[pos][current_index].second != ""){
    if (players[pos][current_index].first == jersey_number){
      players[pos][current_index] = {jersey_number, name};
      return;
    }
    current_index++;

  }
  if (current_index == players[pos].size()){
    vector<pair <int,string> > temp = players[pos];
    players[pos] = vector<pair <int,string> > (players[pos].size()*2);
    for (int i=0 ; i<temp.size() ; i++) {
      players[pos][i] = temp[i];
    }
  }
  players[pos][current_index] = {jersey_number, name};
}


void delete_seperate_chaining(int jersey_number){
  int pos = hashjersey(jersey_number);
  int current_index = 0;
  int position_to_swap = -1;

  while (current_index < players[pos].size() && players[pos][current_index].second != "" ){
    if (players[pos][current_index].first == jersey_number){
      players[pos][current_index] = {-1 , ""};
      position_to_swap = current_index;
    }
    current_index++;
  }

  if (position_to_swap == -1){
    return;
  }

  players[pos][position_to_swap] = players[pos][current_index-1];
  if (current_index - 1 != position_to_swap){
    players[pos][current_index-1] = {-1 , ""};
  }
  }


string search_seperate_chaining(int jersey_number){
  int pos = hashjersey(jersey_number);
  int current_index = 0;

  while (current_index < players[pos].size() && players[pos][current_index].second != ""){
    if (players[pos][current_index].first == jersey_number){
      return players[pos][current_index].second;
    }
    current_index++;
  }
  return "Not Found";
  }


int main(){
  insert_seperate_chaining(30,"Dell Curry");
  std::cout << search_seperate_chaining(30) << std::endl;

  insert_seperate_chaining(30,"Stephen Curry");
  std::cout << search_seperate_chaining(30) << std::endl;

  insert_seperate_chaining(3,"Allen Iverson");
  insert_seperate_chaining(13,"James Harden");
  insert_seperate_chaining(23,"Michael Jordan");
  std::cout << players[hashjersey(23)].size() << std::endl;
  std::cout << search_seperate_chaining(33) << std::endl;
  insert_seperate_chaining(33,"Larry Bird");
  std::cout << search_seperate_chaining(33) << std::endl;
  insert_seperate_chaining(43,"Pascal Siakam");
  std::cout << players[hashjersey(23)].size() << std::endl;

  delete_seperate_chaining(13);
  std::cout << search_seperate_chaining(33) << std::endl;
  std::cout << search_seperate_chaining(43) << std::endl;

  delete_seperate_chaining(33);
  std::cout << search_seperate_chaining(3) << std::endl;
  std::cout << search_seperate_chaining(23) << std::endl;
  std::cout << search_seperate_chaining(43) << std::endl;

  insert_seperate_chaining(13,"James Harden");
  int hashed_thirteen = hashjersey(13);
  for (int i=0 ; i < players[hashed_thirteen].size() ; i++){
    std::cout << players[hashed_thirteen][i].first << ": " << players[hashed_thirteen][i].second << std::endl;
  }



  delete_seperate_chaining(67);
  std::cout << search_seperate_chaining(67) << std::endl;

  delete_seperate_chaining(30);
  std::cout << search_seperate_chaining(30) << std::endl;

}
```
